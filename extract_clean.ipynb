```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("reports/topics_meta/topic_prevalence_lda.csv")

topic_labels = {
    0: "Responsible Supply Chain & Labor Ethics",
    1: "Water Stewardship & Resource Use",
    2: "Carbon Accounting & Emissions Reporting",
    3: "Data Centers & Information Integrity",
    4: "Innovation & Climate Technologies",
    5: "AI Infrastructure & Biodiversity Impact",
    6: "Corporate ESG Standards & Disclosure",
    7: "Supply Chain Decarbonization",
    8: "Climate Governance & Leadership",
    9: "Community Solar & Environmental Projects",
    10: "Net-Zero & GHG Commitments",
    11: "Green Buildings & Circular Design"
}

pivot = (
    df.pivot_table(index="year", columns="topic_id", values="share", aggfunc="sum")
    .fillna(0)
    .rename(columns=topic_labels)
)

#  Bigger figure + balanced font sizes
fig, ax = plt.subplots(figsize=(16, 9))   # wider 16:9 aspect ratio

pivot.plot.area(ax=ax, alpha=0.85, colormap="tab20")

ax.set_title(
    "Meta ESG Topic Trends Over Time (LDA-derived Share by Year)",
    fontsize=18,
    weight="bold",
    pad=20
)
ax.set_ylabel("Proportion of Pages", fontsize=14)
ax.set_xlabel("Year", fontsize=14)
ax.tick_params(axis='both', which='major', labelsize=12)

#  Legend outside plot, in multiple columns
ax.legend(
    title="ESG Topics",
    title_fontsize=13,
    fontsize=11,
    bbox_to_anchor=(1.25, 0.5),
    loc="center left",
    ncol=1,
    frameon=False
)

plt.tight_layout(pad=3.0)
plt.subplots_adjust(right=0.75)  # give extra space for legend
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the combined topic prevalence file (from joint topic modeling run)
df = pd.read_csv("reports/topics_apple_meta/topic_prevalence_lda.csv")

# Example mapping (adjust to your actual topic-word table from topics_lda.csv)
topic_labels = {
    0: "Decarbonization & Energy Transition",
    1: "Supply Chain Accountability",
    2: "Circular Economy & Materials",
    3: "Digital Infrastructure & AI Efficiency",
    4: "Governance & Disclosure Standards",
    5: "Innovation & Technology for Sustainability",
    6: "Community & Social Engagement"
}

# Replace topic IDs with readable labels
df["topic_label"] = df["topic_id"].map(topic_labels)

# Create side-by-side stacked area charts
companies = df["company"].unique()
fig, axes = plt.subplots(1, len(companies), figsize=(18, 6), sharey=True)

for ax, company in zip(axes, companies):
    sub = df[df["company"] == company]
    pivot = sub.pivot_table(index="year", columns="topic_label", values="share", aggfunc="sum").fillna(0)
    pivot.plot.area(ax=ax, alpha=0.85, colormap="tab20")
    ax.set_title(f"{company} ESG Topic Trends", fontsize=14, weight="bold")
    ax.set_xlabel("Year")
    ax.set_ylabel("Proportion of Pages")
    ax.legend(title="ESG Themes", bbox_to_anchor=(1.05, 1), loc="upper left", fontsize=8, frameon=False)

plt.tight_layout()
plt.show()
```

# 🌍 Event-Driven ESG Narratives and Reactive Corporate Storytelling (2019–2025)

## 📅 Context: Global Shocks and ESG Language Adaptation

Between **2019–2025**, ESG reports by Apple and Meta did not evolve in a vacuum.  
Their language patterns and topic distributions — as captured by our topic modeling — appear to mirror major **global events** and **public discourse shifts**.  

| Year | Global or Sectoral Event | Common ESG Narrative Response |
|------|--------------------------|-------------------------------|
| **2020** | COVID-19 pandemic | Surge in *“community,” “safety,” “resilience,”* and *“well-being”* language, reflecting CSR-oriented storytelling rather than strategic change. |
| **2021** | Social justice and diversity movements | Inclusion of *“equity,” “accessibility,”* and *“inclusion”* sections in reports, emphasizing moral commitment over measurable progress. |
| **2022** | Global ESG regulation surge (IFRS S1/S2, CSRD) | Sharp increase in *“governance,” “disclosure,”* and *“alignment”* terminology — signaling narrative adaptation to regulatory pressure. |
| **2023–2024** | AI boom and tech optimism | Framing sustainability as *innovation* and *AI-driven efficiency*; repositioning ESG around technological competitiveness. |
| **2025 (projected)** | Investor skepticism and disclosure fatigue | ESG discourse becomes *data-heavy yet narratively sanitized* — suggesting a move toward “compliance greenwashing.” |

---

## 📈 Evidence from Topic Trends (Apple vs Meta)

Our topic-prevalence charts reveal several **event-linked shifts** in corporate ESG focus:

| Year | Company | Observed Topic Trend | Likely Narrative Motivation |
|------|----------|----------------------|-----------------------------|
| **2020** | Apple | Rise in *Community & Social Engagement* and *Carbon & Energy* topics | Pandemic-era CSR and operational continuity storytelling. |
| **2020** | Meta | Focus on *Data Centers & Energy Efficiency* | Framing platform continuity during lockdowns as a “social good.” |
| **2021** | Both | Surge in *Governance & Disclosure* topics | Early response to pending global ESG reporting frameworks. |
| **2023–2024** | Meta | Growth in *Innovation & AI for Sustainability* | Piggybacking on AI hype to reframe ESG as technology-led. |
| **2024–2025** | Apple | Spike in *Supplier Decarbonization* and *Scope 3* themes | Anticipation of CSRD and investor-driven accountability demands. |

---

## 🧠 Interpretation: ESG as Reactive Reputation Management

Our temporal topic modeling suggests that both Apple and Meta’s ESG language operates as a **reputation management mechanism**, responding dynamically to the global conversation rather than purely internal sustainability progress.

### Key behavioral patterns:

- **Crisis Response:**  
  During global disruption (2020–2021), ESG discourse pivots toward *humanitarian* and *community* language, consistent with CSR-style crisis framing.

- **Regulatory Mirroring:**  
  The dominance of *Governance & Disclosure* after 2022 coincides with international standardization (IFRS S2, CSRD) — suggesting narrative alignment to compliance agendas.

- **Technological Reframing:**  
  From 2023 onward, both firms increasingly invoke *“innovation,” “AI,”* and *“digital infrastructure efficiency”* — signaling a pivot from traditional environmental stewardship to **techno-optimistic ESG branding**.
