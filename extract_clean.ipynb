```python
from pathlib import Path
import re
import pandas as pd
import numpy as np

# Point this to your data root
DATA_ROOT = Path("reports").resolve()

COMPANIES = ["apple","google","amazon","intel","meta","microsoft","nvidia"]

# Regex helpers to infer company/year from a path or filename
COMP_RE  = re.compile(r"(apple|google|amazon|intel|meta|microsoft|nvidia)", re.I)
YEAR_RE  = re.compile(r"(20(?:1[5-9]|2[0-9]))")  # 2015–2029

def _first_col(df, candidates):
    """Return first matching column name (case-insensitive) from candidates; else None."""
    low = {c.lower(): c for c in df.columns}
    for cand in candidates:
        if cand.lower() in low:
            return low[cand.lower()]
    return None

def _standardize_cols(df):
    df = df.copy()
    # Normalize spaces & lowercase keys
    df.columns = [str(c).strip() for c in df.columns]
    # Try to coerce numeric topic when possible
    for cand in ["topic","topic_id","topicid","label_id","cluster","component"]:
        if cand in df.columns:
            try:
                df[cand] = df[cand].astype(int)
            except Exception:
                pass
            break
    return df

def _extract_company(s: str):
    if not isinstance(s, str): return None
    m = COMP_RE.search(s)
    return m.group(1).lower() if m else None

def _extract_year(s: str):
    if not isinstance(s, str): return None
    m = YEAR_RE.search(s)
    return int(m.group(1)) if m else None

def load_company_frame(company: str) -> pd.DataFrame | None:
    """
    Try per-company folder first: reports/bt_tuning_{company}/best/
    Falls back to aggregated: reports/bt_tuning_all/best/ (and infers company from doc/path).
    Returns long frame with columns: company, topic, count, (optional) year
    """
    # Preferred per-company paths
    base = DATA_ROOT / f"bt_tuning_{company}" / "best"
    prev = base / "topic_prevalence_bertopic.csv"
    docs = base / "doc_topics_bertopic.csv"

    # Fallback aggregated paths
    base_all = DATA_ROOT / "bt_tuning_all" / "best"
    prev_all = base_all / "topic_prevalence_bertopic.csv"
    docs_all = base_all / "doc_topics_bertopic.csv"

    # Helper to convert a prevalence or docs file to standard schema
    def _from_prev(df, force_company=None):
        df = _standardize_cols(df)
        topic_col = _first_col(df, ["topic","topic_id","topicid","label_id","cluster"])
        count_col = _first_col(df, ["count","weight","prob","freq","frequency","value","n"])
        comp_col  = _first_col(df, ["company"])
        year_col  = _first_col(df, ["year","report_year","yr"])
        doc_col   = _first_col(df, ["document","doc","path","filename","source","file"])

        if topic_col is None:
            raise KeyError("No 'topic' column (topic/topic_id/label_id/cluster) found.")
        if count_col is None:
            df["count"] = 1.0
            count_col = "count"

        out = pd.DataFrame({
            "topic": df[topic_col],
            "count": pd.to_numeric(df[count_col], errors="coerce").fillna(0.0)
        })

        # company: prefer explicit, else parse from doc if provided, else use force_company
        if comp_col:
            out["company"] = df[comp_col].astype(str).str.lower()
        elif doc_col:
            out["company"] = df[doc_col].astype(str).apply(_extract_company)
        else:
            out["company"] = force_company

        # year: prefer explicit, else parse from doc if provided
        if year_col:
            out["year"] = pd.to_numeric(df[year_col], errors="coerce")
        elif doc_col:
            out["year"] = df[doc_col].astype(str).apply(_extract_year)
        else:
            out["year"] = np.nan

        out = out.dropna(subset=["company","topic"])
        # Ensure ints where possible
        try: out["topic"] = out["topic"].astype(int)
        except: pass
        return out

    # 1) Per-company prevalence
    if prev.exists():
        return _from_prev(pd.read_csv(prev), force_company=company)

    # 2) Per-company docs
    if docs.exists():
        return _from_prev(pd.read_csv(docs), force_company=company)

    # 3) Aggregated prevalence (must infer company from doc/company cols inside)
    if prev_all.exists():
        return _from_prev(pd.read_csv(prev_all), force_company=None)

    # 4) Aggregated docs
    if docs_all.exists():
        return _from_prev(pd.read_csv(docs_all), force_company=None)

    return None

def load_all_companies() -> pd.DataFrame:
    frames = []
    for c in COMPANIES:
        df = load_company_frame(c)
        if df is not None and not df.empty:
            frames.append(df)
    if not frames:
        raise RuntimeError("Could not find usable prevalence/docs in per-company or 'all' directories.")
    data = pd.concat(frames, ignore_index=True)
    # Final cleanups
    data = data.dropna(subset=["company","topic"])
    data["company"] = data["company"].str.lower()
    return data
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics.pairwise import cosine_similarity

data = load_all_companies()

# Build company × topic matrix
mat = data.groupby(["company","topic"], as_index=False)["count"].sum()
pivot = mat.pivot_table(index="company", columns="topic", values="count", fill_value=0).sort_index()

# Normalize rows to topic *shares* (prevents larger corpora dominating)
row_norm = pivot.div(pivot.sum(axis=1).replace(0, np.nan), axis=0).fillna(0)

sim = cosine_similarity(row_norm.values)
sim_df = pd.DataFrame(sim, index=row_norm.index, columns=row_norm.index)

plt.figure(figsize=(7,6))
sns.heatmap(sim_df, annot=True, fmt=".2f", cmap="coolwarm", linewidths=.5)
plt.title("Cross-Company Topic Similarity (cosine, BERT)", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
from pathlib import Path
import re
import pandas as pd
import numpy as np

DATA_ROOT = Path("reports").resolve()
COMPANIES = ["apple","google","amazon","intel","meta","microsoft","nvidia"]

COMP_RE  = re.compile(r"(apple|google|amazon|intel|meta|microsoft|nvidia)", re.I)
YEAR_RE  = re.compile(r"(20(?:1[5-9]|2[0-9]))")

def _first_col(df, candidates):
    low = {c.lower(): c for c in df.columns}
    for cand in candidates:
        if cand.lower() in low:
            return low[cand.lower()]
    return None

def _standardize_cols(df):
    df = df.copy()
    df.columns = [str(c).strip() for c in df.columns]
    for cand in ["topic","topic_id","topicid","label_id","cluster","component"]:
        if cand in df.columns:
            try: df[cand] = df[cand].astype(int)
            except: pass
            df.rename(columns={cand:"topic"}, inplace=True)
            break
    return df

def _extract_company(s: str):
    if not isinstance(s, str): return None
    m = COMP_RE.search(s)
    return m.group(1).lower() if m else None

def _extract_year(s: str):
    if not isinstance(s, str): return None
    m = YEAR_RE.search(s)
    return int(m.group(1)) if m else None

def _from_prev(df, force_company=None):
    df = _standardize_cols(df)
    topic_col = "topic" if "topic" in df.columns else None
    count_col = _first_col(df, ["count","weight","prob","freq","frequency","value","n"])
    comp_col  = _first_col(df, ["company"])
    year_col  = _first_col(df, ["year","report_year","yr"])
    doc_col   = _first_col(df, ["document","doc","path","filename","source","file"])

    if topic_col is None:
        raise KeyError("No topic column found (topic/topic_id/label_id/cluster).")
    if count_col is None:
        df["count"] = 1.0
        count_col = "count"

    out = pd.DataFrame({
        "topic": df[topic_col],
        "count": pd.to_numeric(df[count_col], errors="coerce").fillna(0.0)
    })

    if comp_col:
        out["company"] = df[comp_col].astype(str).str.lower()
    elif doc_col:
        out["company"] = df[doc_col].astype(str).apply(_extract_company)
    else:
        out["company"] = force_company

    if year_col:
        out["year"] = pd.to_numeric(df[year_col], errors="coerce")
    elif doc_col:
        out["year"] = df[doc_col].astype(str).apply(_extract_year)
    else:
        out["year"] = np.nan

    return out.dropna(subset=["company","topic"])

def load_company_frame(company: str) -> pd.DataFrame | None:
    base = DATA_ROOT / f"bt_tuning_{company}" / "best"
    prev = base / "topic_prevalence_bertopic.csv"
    docs = base / "doc_topics_bertopic.csv"

    base_all = DATA_ROOT / "bt_tuning_all" / "best"
    prev_all = base_all / "topic_prevalence_bertopic.csv"
    docs_all = base_all / "doc_topics_bertopic.csv"

    if prev.exists():     return _from_prev(pd.read_csv(prev), force_company=company)
    if docs.exists():     return _from_prev(pd.read_csv(docs), force_company=company)
    if prev_all.exists(): return _from_prev(pd.read_csv(prev_all))
    if docs_all.exists(): return _from_prev(pd.read_csv(docs_all))
    return None

def load_all_companies() -> pd.DataFrame:
    frames = []
    for c in COMPANIES:
        df = load_company_frame(c)
        if df is not None and not df.empty:
            frames.append(df)
    if not frames:
        raise RuntimeError("No usable prevalence/docs found.")
    data = pd.concat(frames, ignore_index=True).dropna(subset=["company","topic"])
    data["company"] = data["company"].str.lower()
    try: data["topic"] = data["topic"].astype(int)
    except: pass
    return data

def load_topic_info() -> pd.DataFrame | None:
    # prefer 'all', else first company with a file
    p_all = DATA_ROOT / "bt_tuning_all" / "best" / "topics_bertopic_info.csv"
    if p_all.exists(): return pd.read_csv(p_all)
    for c in COMPANIES:
        p = DATA_ROOT / f"bt_tuning_{c}" / "best" / "topics_bertopic_info.csv"
        if p.exists(): return pd.read_csv(p)
    return None
```

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path

DATA = Path("reports")
COMPANIES = ["apple","google","amazon","intel","meta","microsoft","nvidia"]

def load_prevalence():
    dfs=[]
    for c in COMPANIES:
        f=DATA/f"bt_tuning_{c}/best/topic_prevalence_bertopic.csv"
        if f.exists():
            df=pd.read_csv(f)
            df=df.rename(columns={"topic_id":"topic","share":"weight"})
            df["company"]=c
            dfs.append(df[["company","year","topic","weight"]])
    return pd.concat(dfs,ignore_index=True)

counts = load_prevalence()
print(counts)
```

```python
import re
def assign_esg(rep):
    s = str(rep).lower()
    if any(k in s for k in ["carbon","emission","energy","renewable","waste","water","climate"]):
        return "Environment"
    if any(k in s for k in ["employee","community","diversity","safety","labor","inclusion","social"]):
        return "Social"
    if any(k in s for k in ["ethic","governance","board","transparency","privacy","responsib"]):
        return "Governance"
    return "Other"

topic2esg = info.set_index("Topic")["Representation"].apply(assign_esg).to_dict()
counts["esg"] = counts["topic"].map(lambda t: topic2esg.get(t,"Other"))
```

```python
# company-agnostic aggregated ESG emphasis per year
ts = counts.groupby(["year","esg"])["weight"].sum().reset_index()
pivot = ts.pivot(index="year",columns="esg",values="weight").fillna(0)

plt.figure(figsize=(8,4))
sns.heatmap(pivot,annot=False,cmap="YlGnBu",cbar_kws={"label":"Textual ESG intensity"})
plt.title("ESG Emphasis Over Time (All Companies, BERT Topics)")
plt.ylabel("Year"); plt.xlabel("ESG Dimension")
plt.tight_layout()
plt.show()
```
