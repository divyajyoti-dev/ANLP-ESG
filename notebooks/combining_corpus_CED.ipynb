```python
import pandas as pd
import glob

paths = glob.glob("/Users/divyajyoti/dev/grad/NLP/ANLP-ESG/data/processed/*/corpus_pages_clean_final.csv")

dfs = []
for p in paths:
    df = pd.read_csv(p)
    # derive company from folder name
    company = p.split("/")[-2].replace("_layout_all", "")
    df["company"] = company
    dfs.append(df)

corpus = pd.concat(dfs, ignore_index=True)

corpus.to_csv("/Users/divyajyoti/dev/grad/NLP/ANLP-ESG/data/processed/all_corpus_cleaned_final.csv", index=False)
corpus.head(100)
```

```python
import pandas as pd
import nltk
nltk.download('punkt')

df = pd.read_csv("/Users/divyajyoti/dev/grad/NLP/ANLP-ESG/data/processed/all_corpus_cleaned_final.csv")

def split_into_sentences(text):
    if pd.isna(text) or text.strip() == "":
        return []
    return nltk.sent_tokenize(text)

rows = []
for idx, row in df.iterrows():
    sentences = split_into_sentences(row["page_text"])
    
    for i, s in enumerate(sentences):
        rows.append({
            "company": row["company"],
            "report_name": row["report_name"],
            "year": row["year"],
            "page": row["page"],
            "sentence_id": i,
            "sentence_text": s.strip()
        })

sent_df = pd.DataFrame(rows)
sent_df.to_csv("/Users/divyajyoti/dev/grad/NLP/ANLP-ESG/data/processed/all_corpus_sentences.csv", index=False)

sent_df.tail()
```
